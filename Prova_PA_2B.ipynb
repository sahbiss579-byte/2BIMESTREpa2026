```python
# Questão 2
n = int(input("Digite um número: "))
soma = 0
processo = ""

for i in range(1, n + 1):
    soma += i
    if i == n:
        processo += str(i)
    else:
        processo += str(i) + " + "

print(f"{processo} = {soma}")
```

```python
# Questão 3
numero = int(input("digite o número: "))
print(f"Tabuada do {numero}:")
for i in range(1, 11):
    resultado = numero * i
    print(f"{numero} x {i} = {resultado}")
```

```python
# Questão 4
n = int(input("Digite um número: "))

if n < 0:
    print("Por favor, digite um número positivo.")
else:
    fatorial = 1
    processo = ""

    # Construindo o processo e calculando
    for i in range(n, 0, -1):
        fatorial *= i
        if i == 1:
            processo += str(i)
        else:
            processo += str(i) + " × "

    print(f"{n}! = {processo}")
    print(f"Fatorial = {fatorial}")
```

```python
# Questão 5
valor = int(input("Digite o valor do saque: R$ "))
notas = [100, 50, 20, 10, 5, 2, 1]

for nota in notas:
    quantidade = 0
    while valor >= nota:
        valor -= nota
        quantidade += 1
    if quantidade > 0:
        print(f"{quantidade} nota(s) de R${nota}")
```
