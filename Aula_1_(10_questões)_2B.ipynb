```
# Atividade 2: Contagem Regressiva (10 a 0)
for i in range(10, -1, -1):
    print(i)
```

```
# Atividade 3: Números Pares
numero = int(input("digite um numero: "))
for i in range(numero + 1):
    if i % 2 == 0:
        print(i)
```

```
# Atividade 4: Números Ímpares
numero = int(input("digite um numero: "))
for i in range(numero + 1):
    if i % 2 != 0:
        print(i)
```

```
# Atividade 5: Soma dos números de 0 a 10
soma = 0
for i in range(11):
    soma += i
print(f"A soma dos números de 0 a 10 é: {soma}")
```

```
# Atividade 6
numero = int(input("digite o número: "))
print(f"Tabuada do {numero}:")
for i in range(1, 11):
    resultado = numero * i
    print(f"{numero} x {i} = {resultado}")
```

```
# Atividade 7
palavra = input("digite uma palavra: ")
numero_letras = len(palavra)
print(f"A palavra '{palavra}' tem {numero_letras} letras.")
```

```
# Atividade 8
palavra = input("digite uma palavra: ")
palavra = palavra.lower() # Converte a palavra para minúsculas para facilitar a contagem de vogais

vogais = "aeiou"
numero_vogais = 0

for letra in palavra:
    if letra in vogais:
        numero_vogais += 1

print(f"A palavra '{palavra}' tem {numero_vogais} vogais.")
```

```
# atividade 9
numeros = []

# Faz um laço para pedir 4 números e armazena na lista
for i in range(1, 5):
    num = float(input(f"Digite o {i}º número: "))
    numeros.append(num)

    # A função max() encontra o maior valor dentro da lista
    maior_numero = max(numeros)

    print(f"O maior número digitado foi: {maior_numero}")
```

```
# Atividade 10
numero = int(input("Digite um número entre 1 e 100: "))

if 1 <= numero <= 100:
    print(f"\nNúmeros entre 1 e {numero} que são divisíveis por 3 e 5 ao mesmo tempo:")
    for i in range(1, numero + 1):
        if i % 3 == 0 and i % 5 == 0:
            print(i)
else:
    print("Número inválido! Por favor, digite um número que esteja entre 1 e 100.")
```
